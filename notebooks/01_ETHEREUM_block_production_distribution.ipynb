```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/raw/ethereum_blocks.csv")
df.head()

proposer_col = "proposer_pubkey" if "proposer_pubkey" in df.columns else "proposer"
counts = df[proposer_col].value_counts()
counts
```

```python
counts.plot(kind="bar")
plt.title("Block production distribution")
plt.xlabel("Proposer")
plt.ylabel("Number of blocks")
plt.show()
```

```python
top_1 = counts.iloc[0] / counts.sum()
top_3 = counts.iloc[:3].sum() / counts.sum()
top_10 = counts.iloc[:10].sum() / counts.sum()

top_1, top_3, top_10
```
