```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/raw/blocks.csv")
df.head()
counts = df["proposer"].value_counts()
counts
```

```python
counts.plot(kind="bar")
plt.title("Blocks produced by Proposer (linea)")
plt.xlabel("Proposer")
plt.ylabel("Blocks")
plt.show()
```

```python
top_1 = counts.iloc[0] / counts.sum()
top_3 = counts.iloc[:3].sum() / counts.sum()
top_5 = counts.iloc[:5].sum() / counts.sum()

top_1, top_3, top_5
```
